## Encontrando Flags (Finding Flags)

Cada desafio foi criado para te ajudar a entender novos conceitos de criptografia. Para resolver, você precisará encontrar uma "flag".

As flags geralmente têm o formato `crypto{y0ur_f1rst_fl4g}`, o que facilita verificar se você acertou.

---

**Por que usar Flags?**

Flags são uma forma divertida de validar seu progresso e garantir que você está no caminho certo. Elas também servem como uma prova de que você superou um desafio específico.

Na vida real, ao invés de uma flag, você pode encontrar tokens de autenticação ou chaves criptográficas. Como este é um ambiente de aprendizado, usamos flags para simplificar e garantir que você acessou informações ou recursos importantes.

---

**Execute a célula abaixo e insira a flag:** `crypto{y0ur_f1rst_fl4g}`


```python
from IPython.display import display, HTML
import base64

flag = "Y3J5cHRve3kwdXJfZjFyc3RfZmw0Z30="
flag_usuario = input('Insira a flag: ')

flag_certa = base64.b64decode(flag).decode('utf-8')

if flag_usuario == flag_certa:
    display(HTML('<span style="color: green; font-size: 32px; font-weight: bold;">✔️ Parabéns, você entendeu a dinâmica!!</span>'))
else:
    display(HTML('<span style="color: red; font-size: 32px; font-weight: bold;">❌ A flag deve ser fornecida com exatidão.</span>'))
```
