## ASCII (5 pts)

**ASCII** (American Standard Code for Information Interchange) é um padrão de codificação de caracteres de 7 bits que permite a representação de texto usando números inteiros de 0 a 127.

Usando a lista de números inteiros abaixo, converta os números para seus caracteres ASCII correspondentes para obter a flag.

`[99, 114, 121, 112, 116, 111, 123, 65, 83, 67, 73, 73, 95, 112, 114, 49, 110, 116, 52, 98, 108, 51, 125]`

Em Python, a função `chr()` pode ser usada para converter um número ordinal ASCII em um caractere (a função `ord()` faz o oposto).

---

### O que é ASCII?

Imagine que você precisa que computadores diferentes entendam e mostrem texto da mesma forma. Para isso, foi criado um "acordo" ou "padrão" chamado ASCII.

*   **Padrão de Codificação:** ASCII é basicamente uma tabela que associa números a caracteres.
*   **7 bits:** Originalmente, usava 7 bits, o que permitia representar 128 caracteres diferentes (2<sup>7</sup> = 128). Estes incluem:
    *   Letras maiúsculas (A-Z)
    *   Letras minúsculas (a-z)
    *   Números (0-9)
    *   Símbolos de pontuação (como !, ?, ., ,)
    *   Caracteres de controle (como nova linha, tabulação, que não são visíveis mas afetam a formatação do texto).
*   **Números de 0 a 127:** Cada um desses 128 caracteres tem um número único associado a ele.

**Exemplo Simples:**

*   O número `65` em ASCII representa a letra `'A'`.
*   O número `97` em ASCII representa a letra `'a'`.
*   O número `48` em ASCII representa o dígito `'0'`.

Você pode consultar uma "Tabela ASCII" online para ver todas as correspondências.

---

### Conversão em Python

Python torna essa conversão muito fácil com funções embutidas:

*   **`chr(numero)`:** Recebe um número inteiro (o código ASCII) e retorna o caractere correspondente.
    *   Exemplo: `chr(65)` retorna `'A'`.
*   **`ord(caractere)`:** Faz o oposto. Recebe um caractere e retorna o número inteiro (código ASCII) correspondente.
    *   Exemplo: `ord('A')` retorna `65`.

Neste desafio, você receberá uma lista de números. Sua tarefa é converter cada número para seu caractere ASCII usando `chr()` e depois juntar todos os caracteres para formar a flag.

---

**Execute a célula de código abaixo para obter a flag.**

```python
numeros = [99, 114, 121, 112, 116, 111, 123, 65, 83, 67, 73, 73, 95, 112, 114, 49, 110, 116, 52, 98, 108, 51, 125]

flag_decodificada = ''.join(chr(numero) for numero in numeros)

print(flag_decodificada)
```

```python
# Exercício para determinar performaces das soluções.

import time

# Criamos uma lista de 100 mil números ASCII do caractere 'a' (que é 97)
numeros = [97] * 900_000


# Medindo tempo com +=
inicio_mais = time.time()

resultado_mais = ''

for numero in numeros:
    resultado_mais += chr(numero)
    
fim_mais = time.time()




# Medindo tempo com append
inicio_append = time.time()

flag_caracteres = []

# Itera sobre cada número na lista
for numero in numeros:
    # Converte o número para seu caractere ASCII correspondente
    caractere = chr(numero)
    flag_caracteres.append(caractere)

# Junta todos os caracteres para formar a string da flag
flag_final = "".join(flag_caracteres)

fim_append = time.time()




# Medindo tempo com join
inicio_join = time.time()

resultado_join = ''.join(chr(numero) for numero in numeros)

fim_join = time.time()




# Exibindo os resultados
print(f"Tempo com '+='      : {fim_mais - inicio_mais:.3f} segundos")
print(f"Tempo com append()  :  {fim_append - inicio_append:.3f} segundos")
print(f"Tempo com join()    :  {fim_join - inicio_join:.3f} segundos")
```
